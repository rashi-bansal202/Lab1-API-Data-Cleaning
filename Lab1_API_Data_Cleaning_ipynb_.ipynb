```python

import requests
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

Path("data").mkdir(exist_ok=True)
Path("reports").mkdir(exist_ok=True)

print("All libraries imported successfully!")


```

```python

url = "https://dummyjson.com/products?limit=100"

response = requests.get(url, timeout=30)
response.raise_for_status()

json_data = response.json()
df = pd.DataFrame(json_data["products"])

print("RAW DATA")
display(df.head(10))

print("Dataset shape:", df.shape)
print("Columns:", df.columns.tolist())
```

```python

df.to_csv("raw_products.csv", index=False)

print("Missing values:")
display(df.isnull().sum().to_frame("Missing Count"))

print("Duplicate rows:", df.duplicated().sum())

print("Data types:")
display(df.dtypes.to_frame("Data Type"))
```

```python

columns = ["id", "title", "category", "price", "rating", "stock"]

clean_df = df[columns].copy()

# Create sample issues for demonstrating cleaning
clean_df.loc[0, "price"] = np.nan
clean_df = pd.concat(
    [clean_df, clean_df.iloc[[1]]],
    ignore_index=True
)

print("Before cleaning:")
print("Missing prices:", clean_df["price"].isna().sum())
print("Duplicate rows:", clean_df.duplicated().sum())

# Standardize text
for col in ["title", "category"]:
    clean_df[col] = clean_df[col].astype("string").str.strip()

# Convert numeric columns
for col in ["price", "rating", "stock"]:
    clean_df[col] = pd.to_numeric(clean_df[col], errors="coerce")

# Remove duplicates
clean_df = clean_df.drop_duplicates()

# Fill missing prices using the median
clean_df["price"] = clean_df["price"].fillna(
    clean_df["price"].median()
)

print("\nAfter cleaning:")
print("Missing prices:", clean_df["price"].isna().sum())
print("Duplicate rows:", clean_df.duplicated().sum())

display(clean_df.head())
```

```python

clean_df["valid_price"] = (
    clean_df["price"].notna() & (clean_df["price"] > 0)
)

clean_df["valid_rating"] = clean_df["rating"].between(0, 5)

clean_df["valid_stock"] = (
    clean_df["stock"].notna() & (clean_df["stock"] >= 0)
)

print("Invalid prices:", (~clean_df["valid_price"]).sum())
print("Invalid ratings:", (~clean_df["valid_rating"]).sum())
print("Invalid stock values:", (~clean_df["valid_stock"]).sum())

display(clean_df[
    ["title", "price", "rating", "stock",
     "valid_price", "valid_rating", "valid_stock"]
].head(10))
```

```python

Q1 = clean_df["price"].quantile(0.25)
Q3 = clean_df["price"].quantile(0.75)
IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = clean_df[
    (clean_df["price"] < lower_limit) |
    (clean_df["price"] > upper_limit)
]

print("Lower limit:", lower_limit)
print("Upper limit:", upper_limit)
print("Number of outliers:", len(outliers))

display(outliers[["title", "price"]])
```

```python

plt.figure(figsize=(8, 5))
plt.hist(clean_df["price"], bins=15, edgecolor="black")
plt.title("Distribution of Product Prices")
plt.xlabel("Price")
plt.ylabel("Number of Products")
plt.tight_layout()
plt.savefig("price_distribution.png")
plt.show()
```

```python

plt.figure(figsize=(8, 5))
plt.boxplot(clean_df["price"].dropna())
plt.title("Product Price Outlier Detection")
plt.ylabel("Price")
plt.tight_layout()
plt.savefig("price_outliers.png")
plt.show()
```
